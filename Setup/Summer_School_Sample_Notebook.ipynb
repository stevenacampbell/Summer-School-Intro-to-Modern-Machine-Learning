```python
import numpy as np

x = np.array([1, 2, 3, 4, 5])
print("x =", x)
print("x squared =", x**2)
```

```python
import pandas as pd
from IPython.display import display

table = pd.DataFrame({
    "x": x,
    "x squared": x**2,
})

display(table)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(5, 3))
plt.plot(x, x**2, marker="o")
plt.title("A simple plot")
plt.xlabel("x")
plt.ylabel("x squared")
plt.grid(True)
plt.show()
```
